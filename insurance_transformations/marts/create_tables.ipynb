```
%%sql -r dataframe_2
CREATE OR REPLACE TABLE INSURANCE_PROJECT.MARTS.CLAIMS_INFORMATION AS
    SELECT
       f.policy_key,
       f.claim_id,
       f.claim_status,
       t.claim_type,
       f.claim_amount,
       f.fraud_score
    FROM INSURANCE_PROJECT.CORE.FACT_CLAIM f
    LEFT JOIN INSURANCE_PROJECT.CORE.DIM_CLAIM_TYPE t
      ON f.claim_type_key = t.claim_type_key;
```

```
%%sql -r dataframe_1
CREATE OR REPLACE TABLE INSURANCE_PROJECT.MARTS.POLICY_INFORMATION AS
    SELECT
       policy_key,
       customer_key,
       policy_id,
       policy_type,
       policy_status,
       annual_premium,
       deductible,
       sales_channel
    FROM INSURANCE_PROJECT.CORE.DIM_POLICY;
```

```
%%sql -r dataframe_3
CREATE OR REPLACE TABLE INSURANCE_PROJECT.MARTS.CUSTOMER_INFORMATION AS
    SELECT
       customer_key,
       customer_id,
       occupation,
       city,
       region
    FROM INSURANCE_PROJECT.CORE.DIM_CUSTOMER;
```

```
%%sql -r dataframe_4
CREATE OR REPLACE TABLE INSURANCE_PROJECT.MARTS.PAYMENTS_INFORMATION AS

WITH payment_summarized As(
     SELECT
        payment_id,
        payment_method,
        SUM(payment_amount) AS total_paid,
        COUNT(*) AS number_of_payments
     FROM INSURANCE_PROJECT.CORE.FACT_PAYMENT
     GROUP BY payment_id, payment_method
)

SELECT  *
FROM  payment_summarized;
```

```
%%sql -r dataframe_5
CREATE OR REPLACE TABLE INSURANCE_PROJECT.MARTS.EVENTS_SUMMARY AS

WITH event_summarized As(
     SELECT
        claim_id,
        COUNT(*) AS event_count,
        MIN(event_timestamp) AS first_timestamp,
        MAX(event_timestamp) AS last_timestamp
     FROM INSURANCE_PROJECT.CORE.FACT_CLAIM_EVENT
     GROUP BY claim_id
)

SELECT  *
FROM  event_summarized;
```

```
%%sql -r dataframe_6
CREATE OR REPLACE TABLE INSURANCE_PROJECT.MARTS.DATE_INFORMATION AS
   SELECT
      full_date,
      year,
      month,
      quarter,
      month_name,
      is_weekend
   FROM INSURANCE_PROJECT.CORE.DIM_DATES;
      
```
